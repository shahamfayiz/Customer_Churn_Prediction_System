```python
import pandas as pd
```

```python
# loading csv file to a dataframe using pandas
df = pd.read_csv("C:/Users/shaham fayiz/OneDrive/Desktop/customer_churn_ai/data/Telco-Customer-Churn.csv")
df
```

```python
# removing an unwanted column
df = df.drop(columns=["customerID"])
df
```

```python
df.info()
```

```python
# checking for missing values in every column
df.isna().sum()
```

```python
# checking datatypes
df.dtypes
```

```python
# changing object column to a numeric column
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"],errors="coerce")
```

```python
df.isna().sum()

```

```python
# handling missing values
df["TotalCharges"] = df["TotalCharges"].fillna(df["TotalCharges"].median())
df.isna().sum()
```

```python
df['Churn'].value_counts
```

```python
# converting yes to 1 and no 0
binary_columns = [
    "Partner",
    "Dependents",
    "PhoneService",
    "PaperlessBilling",
    "Churn"
]

for column in binary_columns:
    df[column] = df[column].map({
        "Yes": 1,
        "No": 0
    })
df1 = df.copy()
df1.head()
```

```python
# perform one hot encoding
df1 = pd.get_dummies(
    df1,
    drop_first=True,
    dtype=int
)
```

```python
df1.head()
```

```python
x = df1.drop(columns=["Churn"],axis=1)

y = df1["Churn"]

y

```

```python
# perform train/test split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
y_train
```

```python
# perform scalling 
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

X_test_scaled
```

```python
X_test_scaled
```

```python
# building model in logistic regression

from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression(
    max_iter=1000
)

logistic_model.fit(
    X_train_scaled,
    y_train
)

logistic_predictions = logistic_model.predict(
    X_test_scaled
)
```

```python
# build model in randomforestclassifier algorithm
from sklearn.ensemble import RandomForestClassifier

random_forest_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

random_forest_model.fit(
    X_train,
    y_train
)

rf_predictions = random_forest_model.predict(
    X_test
)
```

```python
# evaluate both model performs
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

print("Logistic Regression")

print(
    "Accuracy:",
    accuracy_score(y_test, logistic_predictions)
)

print(
    "Precision:",
    precision_score(y_test, logistic_predictions)
)

print(
    "Recall:",
    recall_score(y_test, logistic_predictions)
)

print(
    "F1 Score:",
    f1_score(y_test, logistic_predictions)
)
print("Random Forest")

print(
    "Accuracy:",
    accuracy_score(y_test, rf_predictions)
)

print(
    "Precision:",
    precision_score(y_test, rf_predictions)
)

print(
    "Recall:",
    recall_score(y_test, rf_predictions)
)

print(
    "F1 Score:",
    f1_score(y_test, rf_predictions)
)
```
